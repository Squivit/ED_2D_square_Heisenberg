```python
import numpy as np
import matplotlib.pyplot as plt
from eta_lattice import SpinConfiguration
from tqdm import tqdm
```

```python
d_range = np.linspace(1., 2., 2)
ts = []
ns = []

for d in tqdm(d_range):
    n = 1.
    while n > .75:
        sc = SpinConfiguration(4, 4, 8, d, .0, 1., 2, False, 1e-8, 0.)
        n = float(sc.get_n())
    ts.append(sc.get_t().real)
    ns.append(n)
```

```python
print(ns)
plt.scatter(d_range, ns)
plt.show()

print(ts)
plt.scatter(d_range, ts)
plt.show()
```
