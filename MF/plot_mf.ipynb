```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm

from lattice_mf_full import SpinConfiguration
```

```python
def plot_mf(steps, mfs, ts, ens, nx = 4, show = True):
    plt.subplot(1, 2, 1)
    plt.scatter(range(steps), mfs, label = f'n', c='purple')
    plt.scatter(range(steps), ts, label = f't', c = 'orange')
    plt.hlines([.5, 0.055], 0, steps-1, ls = '--', colors='gray')
    plt.legend()
    plt.xlabel('Iteration #')

    plt.subplot(1, 2, 2)
    plt.scatter(range(steps), ens, label = r'$\varepsilon$', c = 'green')

    if nx == 4:
        plt.hlines([-0.70178], 0, steps-1, ls = '--', colors='gray')
    elif nx == 6:
        plt.hlines([-0.6896886], 0, steps-1, ls = '--', colors='gray')

    plt.xlabel('Iteration #')
    
    #plt.yscale('log')
    if show:
        plt.show()
```

```python
steps = 50

nx, ny = 4, 4
N = nx * ny
sc = SpinConfiguration(nx, ny, magnon_number=int(N/2)+0, lowest_eignstates=1,
                    delta=1., lamb = 1., k=np.array([0., 0.]), print_data=False)

mfs, ts, ens = sc.go_sc_mf(num_steps = steps, running_params=True)
plot_mf(steps, mfs, ts, ens, nx)
```
