```python
import numpy as np
import matplotlib.pyplot as plt
from pyscmf import get_flow_map_mp, plot_flow_map, plot_polar_dir
```

```python
lambdas = [0., .25, .5, .6, .7, .8, .9, 1.]

delta = 1.
n_bins = 201
t_bins = 201
k_bins = 1000

plot_dir = True

n_rows = 2
n_plots = len(lambdas) + int(plot_dir)
n_cols = int(np.ceil(n_plots/n_rows))

fig = plt.figure(figsize=(12, 5))

for il, l in enumerate(lambdas):
    plt.subplot(n_rows, n_cols, il + 1)
    flow_map = np.loadtxt(f'flow_maps/HF_d_{float(round(delta, 2))}_l_{float(round(l, 4))}_n_{n_bins}_t_{t_bins}_k_{k_bins}.txt')
    ns = np.linspace(0., 1., n_bins)
    ts = np.linspace(0., 1., t_bins)
    plot_flow_map(np.flip(flow_map, axis = 0), ns, ts, show_dir_plot=False)
    plt.yticks([0., .25, .5, .75, 1.], [0., .25, .5, .75, 1.])

    # hide y-axis on not first in the row images
    if il%n_cols > 0:
        plt.yticks([0., .25, .5, .75, 1.], [])
        plt.ylabel('')

    # hide x-axis on non-last row
    if il < n_cols * (n_rows-1):
        plt.xticks([0., .25, .5, .75, 1.], [])
        plt.xlabel('')

    plt.title(rf'$\lambda$ = {float(round(l, 4))}')
    plt.minorticks_on()

ax = plt.subplot(n_rows, n_cols, n_plots)
plot_polar_dir(ax)

plt.show()
```
