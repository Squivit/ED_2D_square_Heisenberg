```python
import numpy as np
import matplotlib.pyplot as plt
from tqdm import tqdm
```

```python
def disp(delta = 1., l = 0., _print = True, lamb_0 : bool = False, mf : bool = False):
    finite_size_parameters = {
        (1.0, 0.0): [-.2206, .05228, 0.04],
        (1.0, 1.0): [-.45206, .5, 0.68],
        (2.0, 0.0): [-.1205, .0148, 0.01],
        (2.0, 1.0): [-.1667, .5, 0.855],
    }

    mf_parameters = {
        #(1.0, 1.0): [-.266954, .0768201, 0.0468892],
        (1.0, 1.0): [-0.2754248488930113, 0.08212106368828764, 0.05072706736003926],
        (2.0, 1.0): [-.139202, .019800, 0.0104697],
    }

    if mf and l != 0.:
        bibj, n, mag_int = mf_parameters[(float(delta), float(l))]
    elif l == 0 or lamb_0:
        bibj, n, mag_int = finite_size_parameters[(float(delta), 0.0)]
    elif l == 1:
        bibj, n, mag_int = finite_size_parameters[(float(delta), float(l))]
    
    t = -bibj

    ak = delta + t - n * (t/2 + l * delta)
    bk = 1 - 2 * n + mag_int + t/2 * (t/2 + l * delta)

    if _print:
        print(rf'D={delta}, l={l}')
        print(f'Ak = {round(ak, 5)}, Bk = {round(bk, 5)}')

        if bk**2 > ak**2:
            print('IMAGINARY')

        gap = 2 * np.sqrt(np.abs(ak**2 - bk**2))
        print(f'Gap at G point: {round(gap, 4)}')
        #else:
        #    print(ak)
        #    print(bk)
        
        print(f'Gap at X point: {round(2 * np.sqrt(ak**2), 4)}')

        #print(t, n, mag_int)

    return ak, bk

disp(delta = 1, l = 0)
disp(delta = 1, l = 1, mf = True)
disp(delta = 1, l = 1, lamb_0 = False)
disp(delta = 1, l = 1, lamb_0 = True)
print()
#disp(delta = 2, l = 0)
#disp(delta = 2, l = 1, mf = True)
#disp(delta = 2, l = 1, lamb_0 = False)
#disp(delta = 2, l = 1, lamb_0 = True)
```
